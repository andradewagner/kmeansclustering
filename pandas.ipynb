```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url_dados = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
```

```python
attributes = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']

iris = pd.read_csv(url_dados, names=attributes)

type(iris)
```

```python
plt.scatter(iris['sepal_length'], iris['sepal_width'])
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
```

```python
colours = {'Iris-setosa': 'red', 'Iris-versicolor': 'blue', 'Iris-virginica': 'green'}
for i in range(len(iris['sepal_length'])):
    plt.scatter(iris['sepal_length'][i], iris['sepal_width'][i], color=colours[iris['class'][i]])
plt.title('Sepal Length vs Sepal Width by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
```
